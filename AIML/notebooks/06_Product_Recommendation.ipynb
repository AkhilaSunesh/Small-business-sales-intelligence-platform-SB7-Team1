```python
import pandas as pd
```

```python
df = pd.read_csv("../data/preprocessed_data.csv")

df.head()
```

```python
df["ProductID"].value_counts()
```

```python
product_counts = (
    df["ProductID"]
    .value_counts()
    .reset_index()
)

product_counts.columns = [
    "ProductID",
    "PurchaseCount"
]

product_counts
```

```python
def recommend_products(product_id, top_n=3):

    recommendations = product_counts[
        product_counts["ProductID"] != product_id
    ].head(top_n)

    return recommendations
```

```python
product_counts.to_csv(
    "../data/product_recommendations.csv",
    index=False
)

print("Recommendation data saved successfully!")
```
