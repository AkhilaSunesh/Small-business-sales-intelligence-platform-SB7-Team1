```python
import pandas as pd
import numpy as np
```

```python
# LOAD DATASET

df = pd.read_csv("../data/Retail_Transaction_Dataset.csv")

df.head()
```

```python
# DATASET SHAPE

print("Shape of Dataset:", df.shape)
```

```python
# DATASET INFORMATION

df.info()
```

```python
# MISSING VALUES

print(df.isnull().sum())
```

```python
# DUPLICATE RECORDS

print("Duplicate Rows =", df.duplicated().sum())
```

```python
# REMOVE DUPLICATE RECORDS

df.drop_duplicates(inplace=True)

print("Shape after removing duplicates:", df.shape)
```

```python
# CONVERT DATE COLUMN


df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], format="mixed")

print("Date Converted Successfully!")
```

```python
# CREATE DATE FEATURES

df["Year"] = df["TransactionDate"].dt.year
df["Month"] = df["TransactionDate"].dt.month
df["Day"] = df["TransactionDate"].dt.day
df["Week"] = df["TransactionDate"].dt.isocalendar().week.astype(int)
df["Quarter"] = df["TransactionDate"].dt.quarter
df["DayOfWeek"] = df["TransactionDate"].dt.day_name()
df["Weekend"] = df["TransactionDate"].dt.dayofweek >= 5

df.head()
```

```python
# CREATE SEASON FEATURE

def get_season(month):

    if month in [12,1,2]:
        return "Winter"

    elif month in [3,4,5]:
        return "Summer"

    elif month in [6,7,8]:
        return "Monsoon"

    else:
        return "Autumn"

df["Season"] = df["Month"].apply(get_season)

df.head()
```

```python
# DATASET INFORMATION AFTER PREPROCESSING

df.info()
```

```python
# PREPROCESSED DATASET

df.head()
```

```python
# SAVE PREPROCESSED DATASET

df.to_csv("../data/preprocessed_data.csv", index=False)

print("Preprocessed Dataset Saved Successfully!")
```

```python
## Preprocessing Summary

#- Removed duplicate records.
#- Converted TransactionDate to datetime format.
#- Extracted Year, Month, Day, Week, Quarter and DayOfWeek.
#- Identified Weekend transactions.
#- Created Season feature.
#- Saved the cleaned dataset for forecasting.
```
