```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/Retail_Transaction_Dataset.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
mean_amount = df["TotalAmount"].mean()
std_amount = df["TotalAmount"].std()

upper_limit = mean_amount + 3 * std_amount
lower_limit = mean_amount - 3 * std_amount

print("Mean:", mean_amount)
print("Standard Deviation:", std_amount)
print("Upper Limit:", upper_limit)
print("Lower Limit:", lower_limit)
```

```python
anomalies = df[
    (df["TotalAmount"] > upper_limit) |
    (df["TotalAmount"] < lower_limit)
]

anomalies.head()
```

```python
print("Total Transactions:", len(df))
print("Anomalies Found:", len(anomalies))
```

```python
anomalies.sort_values(
    "TotalAmount",
    ascending=False
).head(10)
```

```python
df["Anomaly"] = np.where(
    (df["TotalAmount"] > upper_limit) |
    (df["TotalAmount"] < lower_limit),
    "Anomaly",
    "Normal"
)

df.head()
```

```python
df.to_csv(
    "../data/anomaly_detection.csv",
    index=False
)

print("Anomaly Detection Data Saved Successfully!")
```
